##Numpy array basics 



##creating array from list 

```python
import numpy as np
arr_1d = np.array([1,2,3,4,5])
print("1D array: ", arr_1d)

arr_2d = np.array([[1,2,3], [4,5,6]])
print("2D array: ", arr_2d)
```

##list vs numpy array


```python
py_list = [1,2,3]
print("python list multiplication", py_list*2)

np_array = np.array([1,2,3]) #element wise muyliplication
print("Pyhton array multiplication", np_array*2)

import time 
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n list operation time: ", time.time()- start)

start=time.time()
np_array = np.arange(1000000) * 2
print("\n numpy  operation time: ", time.time()- start)
```

##creating arry from scratch


```python
zeroes = np.zeros((3,4))
print(zeroes)

ones = np.ones((2,3))
print(ones)

full = np.full((2,2), 7)
print(full)
```

##vector, matrix and tensor

```python
vector = np.array([1,2,3])
print("vector", vector)

matrix = np.array([[1,2,3], [5,7,8]])
print("mayrix", matrix)

tensor = np.array([[[1,2],[3,4]],
                    [[5,6], [7,8]] ])
print("tensor", tensor)
```

##array properties 

```python
arr = np.array([[1,2,3], [4,5,6]])
print("Shape", arr.shape)
print("dimension", arr.ndim)
print("size", arr.size)
print("data type", arr.dtype)
```

##array reshaping

```python
arr = np.arange(12) #arange(start(inclusive), stop(exclusive), step)
print("original array ", arr)

reshaped = arr.reshape((3,4))
print("reshaped array", reshaped)

flattened = reshaped.flatten() #cganges in copy 
print(flattened)

#ravel returns view instead of copy 
raveled = reshaped.ravel()
print("raveled array", raveled)

#transpose
transpose = reshaped.T
print("transposed array", transpose)
```
