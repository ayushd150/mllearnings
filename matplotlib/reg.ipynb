```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips) #lmplot is linear regression
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips,hue='sex',palette='rainbow')
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips,hue='sex',palette='rainbow', markers=['o','v'], scatter_kws={'s':50})
```
