make new col for episode count(feature extraction)

```python
df.loc[2]['Title']
```

```python
def extract_episodes(txt):
    check=False
    data=""
    for i in txt:
        if i==')':
            check = False
            return data
        if i=='(':
            check=True
        if check==True:
            data+=i
    
```

```python
df["Episodes"]=df["Title"].apply(extract_episodes)
```

```python
df["Episodes"]=df["Episodes"].str.replace(" eps ","")
```

```python
df.loc[0]['Episodes']
```

```python
df['Episodes']=df['Episodes'].astype(int)
```

```python
df.loc[0]['Title']
```

```python
def extraction_time(txt):
    check=False
    data=""
    for i in range(len(txt)):
        if txt[i]==')':
            for j in range(i+1,i+20):
                data+=txt[j]
            return data
```

```python
df['Title'].apply(extraction_time)
```
