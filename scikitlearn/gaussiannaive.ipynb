```python
import pandas as pd
```

```python
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df.head()
```

```python
df2 = pd.get_dummies(df[['PLAYER', 'CS']]) #converts categorical variables into one-hot encoded column
```

```python
df2
```

```python
df3 = pd.concat([df,df2], axis=1)
```

```python
df3
```

```python
df4=df3.drop(['PLAYER', 'CS'], axis=1)
```

```python
df4
```

```python
X = df4.drop(['YRS'], axis=1)
```

```python
y = df4['YRS']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=10)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
gnb = GaussianNB()
```

```python
gnb.fit(X_train, y_train)
```

```python
y_pred = gnb.predict(X_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

```python
gnb.score(X_train, y_train)
```

```python
gnb.score(X_test, y_test)
```

add parameters

```python
param_grid = {
    'var_smoothing': [0.01, 0.02, 0.003]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_search = GridSearchCV(gnb, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
```

```python
grid_search.fit(X_train, y_train)
```
