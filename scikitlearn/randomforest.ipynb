```python
import pandas as pd
```

```python
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df.head()
```

```python
df = df.drop(columns = ['PLAYER', 'CS'])
```

```python
X = df.iloc[:, 0:13]
```

```python
y = df.iloc[:, 13]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=17, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier()
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test) #takes features from test set(X_test) and uses trained model to outpout predicted labels
```

```python
rf.score(X_test, y_test) #for classifiers, score returns accuracy, for regressors retiurn R^2 score -> coefficient of determination
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

```python
features = pd.DataFrame(rf.feature_importances_, index=X.columns)
```

```python
rf2 = RandomForestClassifier(
    n_estimators=1000, criterion='entropy', min_samples_split=10, max_depth=14, random_state=42
) #n-estimatores -> np. of trees in the forest, more trees = more stable prediction(less variance), criterion is the metric used to measure qualty of split, entropy is info gain (gini impurity), min_sample_split is the min no. of samples reqd to split node into child nodes, larger values = simpler trees (less overfiitting), max_depth is max depth of each decision tree in forest, limits how many splits the tree can make from root to leaf, prevents overfitting by stopping tree from becoming too deep
```

```python
rf2.fit(X_train, y_train)
```

```python
rf2.score(X_test, y_test)
```

```python
y_pred2 = rf2.predict(X_test)
```

```python
print(classification_report(y_test, y_pred2))
```
